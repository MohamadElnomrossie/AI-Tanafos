```python
from sklearn.datasets import load_breast_cancer
from sklearn.cluster import KMeans
import seaborn as sb
from sklearn.metrics import classification_report,confusion_matrix
import matplotlib.pyplot as plt
```

```python
X,y=load_breast_cancer(return_X_y=True)
```

```python
del y
```

```python
model=KMeans(n_clusters=2)
```

```python
model
```

```python
output=model.fit_predict(X)
```

```python
y
```

```python
inertias=[]
ks=[]
for i in range(X.shape[0]):
    model=KMeans(n_clusters=i+1)
    model.fit(X)
    ks.append(i+1)
    inertias.append(model.inertia_)
```

```python
sb.lineplot(y=inertias[:5],x=ks[:5],markers="*")
plt.grid()
plt.xlabel("Number of Ks")
plt.ylabel("Inertias")
```

```python
model=KMeans(n_clusters=2)
output=model.fit_predict(X)
```

```python
normalized_out=[]
for val in output:
    if val==0:
        normalized_out.append(1)
        continue
    normalized_out.append(0)
```

```python
print(classification_report(y,normalized_out))
```

```python
y
```

```python
sb.heatmap(confusion_matrix(y,normalized_out),annot=True,fmt='.0f')
```
